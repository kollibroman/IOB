```python
import math

def calculate_cos(inputFilePath):
    numbers = []

    with open(inputFilePath, 'r') as file:
        for line in file:
            result = math.cos(float(line.strip()))
            numbers.append(result)
            print(f"The cosine of {line.strip()} is {result}")

    with open('output.txt', 'w') as output_file:
        for number in numbers:
            output_file.write(f"{number}\n")
```

```python
calculate_cos('input.txt')
```
