# Zadanie 1

```python
import cv2 as cv
import numpy as np
from matplotlib import pyplot as plt

plt.rcParams["figure.figsize"] = (18, 10)
image = cv.imread("images/example.jpg")
plt.imshow(cv.cvtColor(image, cv.COLOR_BGR2RGB))

kernel = [
    [-1, -1, -1],
    [-1, 8, -1],
    [-1, -1, -1],
]

kernel = np.asarray(kernel)
filtered_image = cv.filter2D(image, -1, kernel=kernel)
plt.imshow(filtered_image)
```

# Zadanie 2

```python
import cv2 as cv
import numpy as np
from matplotlib import pyplot as plt

plt.rcParams["figure.figsize"] = (18, 10)
image = cv.imread("images/example.jpg")
plt.imshow(cv.cvtColor(image, cv.COLOR_BGR2RGB))

image_float = image.astype(np.float32) / 255.0

matrix = np.array([
    [0.393, 0.769, 0.189],
    [0.349, 0.689, 0.168],
    [0.272, 0.534, 0.131]
])

new_image_colors = image_float @ matrix.T

new_image_colors = np.clip(new_image_colors, 0, 1)
plt.imshow(new_image_colors)
```

# Zadanie 3

```python
import cv2 as cv
import numpy as np
from matplotlib import pyplot as plt

plt.rcParams["figure.figsize"] = (18, 10)
image = cv.imread("images/example.jpg")
plt.imshow(cv.cvtColor(image, cv.COLOR_BGR2RGB))

sum_matrix = np.array([0, 128, 128])

colors_matrix = np.array([
    [ 0.229, 0.587, 0.114 ],
    [ 0.500, -0.418, -0.082 ],
    [ -0.168, -0.331, 0.500 ]
])

ycbr = (image @ colors_matrix.T) + sum_matrix
ycbr = np.clip(ycbr, 0, 255).astype(np.uint8)

#plt.imshow(ycbr)

y_channel = ycbr[:, :, 0]
cr_channel = ycbr[:, :, 1]
cb_channel = ycbr[:, :, 2]
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(20, 8))

ax1.imshow(y_channel, cmap='gray')
ax1.set_title("Składowa Y")
ax1.axis('off')

ax2.imshow(cr_channel, cmap='gray')
ax2.set_title("Składowa Cr")
ax2.axis('off')

ax3.imshow(cb_channel, cmap='gray')
ax3.set_title("Składowa Cb")
ax3.axis('off')

plt.tight_layout()
plt.show()
```
